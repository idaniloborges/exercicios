```python
1# ### Números pares com `range`

# Use um `for` com `range` para exibir todos os números pares de **2 a 20** (inclusive).

for c in range(2, 22, 2):
    print(c)
```

```python
2# ### Tabuada personalizada

# Peça um número ao usuário e use `for` com `range` para imprimir sua tabuada de **1 a 10**.

number = int(input('Digite um número inteiro: '))

for c in range(1, 11, 1):
    print(f'{number} x {c} = {number * c}')
```

```python
3# ### Soma de elementos de duas listas com `zip`

# Dadas duas listas do mesmo tamanho, use `zip` e `for` para criar uma **nova lista** com as somas dos elementos correspondentes.

# Exemplo:
a = [1,2,3]
b = [4,5,6]
# Saída: [5,7,9]

for number1, number2 in zip(a, b):
    print(number1 + number2)

```

```python
4# ### Combinação de nomes e idades com `zip`

# 5 listas com dados de cadastros nome, idade, profissao, endereco hobbie : `
nomes = ['Ana','Pedro','João', 'Marcos', 'Ezequiel']
idades = [20,25,30, 27, 35]
profissoes = ['analista', 'operador', 'balconista', 'medico', 'advogado']
enderecos = [
    "Rua das Flores, 123 - São Paulo, SP",
    "Av. Brasil, 456 - Rio de Janeiro, RJ",
    "Rua das Acácias, 789 - Belo Horizonte, MG",
    "Praça Central, 101 - Curitiba, PR",
    "Rua do Comércio, 202 - Salvador, BA"
]

hobbies = [
    "Ler livros",
    "Tocar violão",
    "Cozinhar",
    "Fazer trilhas",
    "Pintar quadros"
]

# Use `zip` e `for` para imprimir mensagens como:

# `Ana tem 20 anos.`

for nome, idade, profissao, endereco, hobbie in zip(nomes, idades, profissoes, enderecos, hobbies):
    print(f'{nome} tem {idade} anos e é {profissao}. Reside no endereço: {endereco} e seu hobbie favorito é {hobbie}')

```

```python
5# ### Enumerar frutas

# Dada uma lista `frutas = ['maçã','banana','uva']`, use `for` e `enumerate` para imprimir o índice e o nome de cada fruta.

frutas = ['maçã','banana','uva']

for e in enumerate(frutas):
    print(e)
```

```python
6# ### Manipulação de tuplas

# Dada uma lista de tuplas `pessoas = [('Ana',20),('Pedro',25),('João',30)]`, 
# use `for` para imprimir apenas os nomes e coloque se é maior de idade ou não.

pessoas = [('Ana',17),('Pedro',25),('João',30)]
contador = 0
for nome, idade in pessoas:
    if idade >= 18:
        print(f'{nome} é maior de idade')
    else:
        print(f'{nome} é menor de idade')
```

```python
7# ### Contar palavras em arquivo (`with`)

# Crie um arquivo de texto chamado `texto.txt`. Use `with open()` e `for` para contar quantas linhas e palavras o arquivo possui.

contador_linhas = 0
contador_palavras = 0

with open ("arquivo.txt", 'r') as file:
    for line in file:
        contador_linhas += 1
        contador_palavras += len(line.split())
    
    print(contador_palavras)
    print(contador_linhas)
        
        
```

```python
8# ### Dicionário de notas

# Dado um dicionário `notas = {'Ana':8,'Pedro':7,'João':9}`, use `for` para imprimir apenas os nomes dos alunos que tiraram nota maior ou igual a 8.

notas = {'Ana':8,'Pedro':7,'João':9}

for nome, nota in notas.items():
    if nota >= 8:
        print(nome)

```

```python
9# ### Criar dicionário com quadrados de números

# Use um `for` com `range(1,6)` para criar um dicionário em que a **chave** seja o número e o **valor** seja seu quadrado.

# Exemplo: `{1:1, 2:4, 3:9, 4:16, 5:25}`.

for c in range(1, 6, 1):
    print(c, c*c)
```

```python
10# ### Ler pares chave-valor de um arquivo

# Crie um arquivo `dados.txt` com linhas como:

# Ana 8
# Pedro 7
# João 9

c = 0
with open ("dados.txt", 'r') as file:
    for line in file:
        usuario = line.split()
        print(f'chave: {usuario[c]}, valor: {usuario[c+1]}')
        
```
