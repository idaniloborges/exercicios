```python
# ### 1. Senha até acertar

# Crie um programa que peça a senha de acesso (`input`). 
# Enquanto a senha digitada for diferente de `"1234"`, mostre `"Senha incorreta, tente novamente!"`. Quando acertar, exiba `"Acesso permitido"`.

senha = input('Digite sua senha: ')

while senha != '1234':
    print('Senha incorreta.')
    senha = input('Tente novamente: ')
print('Acesso permitido.')
```

```python
# 2. Contador de notas de alunos

# Peça notas de alunos (`input`) e calcule a média. O programa deve parar apenas quando o usuário digitar `-1`. Ao final, exiba a média das notas.

notas = int(input('Digite suas notas: '))
cont = 0
somatorio_notas = 0

while notas != -1:
    somatorio_notas += notas
    cont += 1
    print(somatorio_notas)
    print(cont)
    print(f'Nota digitada: {notas}')
    print(f'Média atual: {(somatorio_notas / cont):.2f}')
    notas = int(input('Digite suas notas: '))
    
print(f'Média final: {(somatorio_notas / cont):.2f}')
```

```python
# 3. Caixa eletrônico simples

# Simule um caixa eletrônico: peça ao usuário para digitar um valor de saque. 
# Enquanto o valor for maior que o saldo disponível (comece com 1000), peça outro valor. 
# Quando o valor for válido, subtraia do saldo e mostre o saldo restante.

saque = float(input('Digite o valor a ser sacado: '))
saldo = 1000

while saque < saldo:
    print(f'Saque realizado no valor de R${saque:.2f}')
    saldo -= saque
    print(f'Saldo restante de R${saldo:.2f}')
    saque = float(input('Digite o valor a ser sacado: '))
print('O valor a ser sacado é maior do que o saldo.')
```

```python
# 4. Acúmulo de compras

# Peça ao usuário o valor das compras. Continue pedindo até que ele digite `0`. Ao final, exiba o total gasto.

compras = float(input('Digite o valor da compra: '))
total_compras = 0

while compras != 0:
    total_compras += compras
    print(f'Total R${total_compras:.2f}')
    compras = float(input('Digite o valor da compra: '))
print(f'Total final das compras R${total_compras:.2f}')
```

```python
# 7. Conversor de temperaturas
# F° = Fahrenheit
# C° = Celsius
# A fórmula é °F = (°C × 1,8) + 32.
# Permita ao usuário converter temperaturas de Celsius para Fahrenheit. O programa deve continuar pedindo valores até que o usuário digite "sair".

temperatura_c = input('Digite a temperatura')
temperatura_f = (float(temperatura_c) * 1.8) + 32

while temperatura_c != 'sair':
    print(f'{float(temperatura_c):.2f}°C equivale a {temperatura_f:.2f}°F')
    temperatura_c = input('Digite a temperatura')
print('PROGRAMA FINALIZADO')
```

```python
# 8. Controle de tentativas de login
# Permita ao usuário tentar fazer login no máximo 3 vezes. 
# Se ele errar todas, exiba "Conta bloqueada". 
# Se acertar antes, exiba "Login realizado com sucesso".

senha = input('Digite sua senha: ')
tentativas = 0

while True:
    tentativas += 1
    if tentativas < 3 and senha == '1234':
        print('Login realizado com sucesso.')
        break
    elif tentativas < 3 and senha != '1234':
        print('Senha incorreta! Tente novamente.')
        senha = input('Digite sua senha: ')
    else:
        print('Conta bloqueada!')
        break
```

```python
# 9. Soma de números positivos
# Peça números inteiros. Some apenas os positivos. Pare quando o usuário digitar um número negativo e mostre o total da soma.

numero = int(input('Digite um número: '))
somatorio = 0

while numero >= 0:
    print(f'O número digitado foi: {numero}')
    somatorio += numero
    numero = int(input('Digite um número: '))
print(f'Total da soma: {somatorio}')
```

```python
# 10. Lista de tarefas
# Permita que o usuário adicione tarefas (input) a uma lista. Continue pedindo até que ele digite "fim".
# Ao final, imprima todas as tarefas cadastradas.

item = input('Digite a tarefa: ')
lista = []
index = 0

while item != 'fim':
    lista.append(item)
    item = input('Digite a tarefa: ')
for item in lista:
    index += 1
    print(f'Tarefa {index}: {item}')
```
