```python
# 1)Faça uma aplicação que pegue diferentes caracteres. 
# E caso seja diferente do alfabeto ( a até z ) lance a exceção type error.
# Para lançar uma exceção utlizar clausula: raise TypeError("Caractere errado").

import re

try:
    entrada = input('Digite textos: ')
    if not re.search("[^a-zA-Z\s]", entrada):
        print('Ok')
    else: 
        raise TypeError('Nok.')
except TypeError:
    print('Caractere errado.')
```

```python
## **Exercício 1 — Função de soma**

# Crie uma função `soma(a, b)` que retorna a soma de dois números.

# Depois, crie testes para validar os seguintes casos:

# ✅ 1. `soma(2, 3)` deve retornar `5`

# ✅ 2. `soma(-2, 2)` deve retornar `0`

# ❌ 3. `soma(10, 5)` **não deve ser igual** a `20`

import unittest

def soma(a,b):
    return a+b

testcase = unittest.TestCase()

try:
    testcase.assertEqual(soma(2, 3),5)
    testcase.assertEqual(soma(-2, 2), 0)
    testcase.assertNotEqual(soma(10, 5), 20)
    print('Todos os teste passaram!')
except AssertionError as e:
    print('Erro de teste: ', e)
```

```python
# ## **Exercício 2 — Função que verifica se é par**

# Crie uma função `eh_par(numero)` que retorna `"Par"` se o número for par, ou `"Ímpar"` caso contrário.

# Teste as seguintes condições:

# ✅ 1. `eh_par(10)` → `"Par"`

# ✅ 2. `eh_par(3)` → `"Ímpar"`

# ❌ 3. `eh_par(5)` **não deve retornar** `"Pa`

def is_even(a):
    if a%2 == 0:
        return 'Par'
    elif a%2 != 0:
        return 'Ímpar'

testcase = unittest.TestCase()

try:
    testcase.assertEqual(is_even(10), 'Par')
    testcase.assertEqual(is_even(5), 'Ímpar')
    testcase.assertNotEqual(is_even(5), 'Par')
    print('Todos os testes passaram!')

except AssertionError as e:
    print('Erro de teste: ', e)
```

```python
# ## **Exercício 3 — Função que inverte texto**

# Crie uma função `inverter_texto(texto)` que retorna o texto invertido.

# Teste as seguintes condições:

# ✅ 1. `"python"` deve se tornar `"nohtyp"`

# ✅ 2. `"abc"` não deve ser igual a `"abc"` (pois deve inverter)

# ✅ 3. Texto vazio `""` deve continuar sendo `""`


texto = input('Digite algo: ')

def inverter_texto(texto):
    texto_invertido = texto[::-1]
    return texto_invertido

testcase = unittest.TestCase()

try:
    testcase.assertEqual(inverter_texto(texto), texto[::-1])
    # testcase.assertNotEqual(inverter_texto(texto), texto)
    testcase.assertEqual(inverter_texto(texto("")), "")
    print('Todos os testes passaram!')

except AssertionError as e:
    print('Erro de teste: ', e)
    
# print(inverter_texto(texto))
```
