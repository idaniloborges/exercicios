```python
### **1. Função para somar todos os elementos de uma lista**

# Crie uma função chamada somar_lista(lista) que retorna a soma de todos os elementos da lista passada como argumento.

lista = [5, 3, 2, 6, 8, 11, 55, 558, 28, 645, 225, 412]

def somar_lista():
    acumulador = 0
    for numero in lista:
        acumulador += numero
    return acumulador

print(f'Resultado da somatória: {somar_lista()}')
print(max(lista))
```

```python
### **2. Função que retorna o maior e o menor valor da lista**

# Crie uma função extremos(lista) que retorna uma tupla com o menor e o maior número da lista.

lista = [5, 3, 2, 6, 8, 11, 55, 558, 28, 645, 225, 412]

def extremos():
    valor_maximo = max(lista)
    valor_minimo = min(lista)
    resultado = (valor_minimo, valor_maximo)
    return resultado

print(f'Valor mínimo: {extremos()[0]}\nValor máximo: {extremos()[1]}')
```
