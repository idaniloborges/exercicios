```python
# ### 1. Criação e escrita inicial
# Crie um arquivo chamado `alunos.txt` e grave nele os nomes de **três alunos** (um em cada linha).
# Depois, abra o arquivo em modo leitura e exiba seu conteúdo na tela.

with open("alunos.txt",'a') as arquivo:
    arquivo.write('João; ')
    arquivo.write('Maria; ')
    arquivo.write('Davi; ')

with open("alunos.txt", 'r') as reader:
    print(reader.read())
```

```python
# ### 2. Acrescentar conteúdo sem apagar o existente

# Abra o arquivo `alunos.txt` no modo **append** (`'a'`) e adicione mais dois nomes de alunos.

# Depois, exiba o conteúdo atualizado do arquivo.

with open("alunos.txt", 'a') as arquivo:
    arquivo.write('Daniel; ')
    arquivo.write('Jeremias; ')

with open("alunos.txt", 'r') as reader:
    print(reader.read())
```

```python
# ### 3. Reescrita total do arquivo

# Abra o arquivo `alunos.txt` no modo **write** (`'w'`) e grave apenas a frase:
# Lista de alunos apagada!

# Depois, abra novamente o arquivo e confirme que os nomes antigos foram apagados.

with open("alunos.txt", 'w') as arquivo:
    arquivo.write('Lista de alunos apagada!')

with open("alunos.txt", 'r') as reader:
    print(reader.read())
```

```python
# ### 4. Renomear arquivo

# Recrie o arquivo `alunos.txt` com qualquer conteúdo.

# Renomeie-o para `turmaA.txt` usando `os.rename()`.

# Depois, leia o conteúdo do novo arquivo `turmaA.txt` e mostre na tela.

import os

os.rename("alunos.txt", "turmaA.txt")
```

```python
# 5. Exclusão de arquivo
# Crie um arquivo chamado temporario.txt e grave a frase:
# Este arquivo será excluído.
# Use os.remove() para excluir o arquivo.
# Depois, tente abrir temporario.txt e trate o erro com um try/except exibindo a mensagem:
# Arquivo não encontrado.

with open("temporario.txt", 'a') as arquivo:
    arquivo.write('Este arquivo será excluído.')

import os

os.remove("temporario.txt")

try:
    with open("temporario.txt", 'r') as reader:
        print(reader.read())

except:
    print('Arquivo não encontrado.')
```
