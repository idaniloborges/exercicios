```python
try:
    a = input('Digite um no: ')
    c = 1 + int(a)
    print(f'Resultado: {c}')
    
except ValueError:
    print('Erro: valor errado.')
    
except NameError:
    print('Erro: Variável errada.')
    
except TypeError:
    print('Erro: Tipo de dado errado.')
    
except:
    print('Erro')
    
finally:
    print('Sempre executa')
```

```python
# 1)Faça uma aplicação que pegue diferentes caracteres. 
# E caso seja diferente do alfabeto ( a até z ) lance a exceção type error.
# Para lançar uma exceção utlizar clausula: raise TypeError("Caractere errado").

import re

try:
    entrada = input('Digite textos: ')
    if not re.search("[^a-zA-Z\s]", entrada):
        print('Ok')
    else: print('Caractere errado.')
except TypeError:
    print('Caractere errado.')
```

```python
# 2) Faça uma aplicação que cada exceção lançada seja guardada em um arquivo log.txt

try:
    a = input('Digite um número: ')
    c = 1 + int(a)
    print(c)
except ValueError:
    print('Valor errado')
    with open("log.txt", 'a') as arquivo:
        arquivo.write("ValueError: sistema com problema na linha 1.")
except NameError:
    print('Variável errada')
    with open("log.txt", 'a') as arquivo:
        arquivo.write("ValueError: sistema com problema na linha 2.")
except TypeError:
    print('Tipo de dado errado')
    with open("log.txt", 'a') as arquivo:
        arquivo.write("ValueError: sistema com problema na linha 3.")
except TimeoutError:
    print('timeout error')
    with open("log.txt", 'a') as arquivo:
        arquivo.write("ValueError: sistema com problema na linha 4.")
except:
    print('Exceção não especificada')
    with open("log.txt", 'a') as arquivo:
        arquivo.write("ValueError: sistema com problema na linha 5.")
```

```python
# 3) Pegue o primeiro exercício e leia as linhas com for. 
# Se vir mais de 3 linhas emita uma mensagem: Tente novamente mais tarde.

import re

try:
    contador = 0
    while contador <= 3:
        entrada = input('Digite textos: ')
        if not re.search("[^a-zA-Z\s]", entrada):
            print('Ok')
            contador += 1
            print(contador)
        else:
            print('Digite um valor válido.')
    print('Tente novamente mais tarde')        
except TypeError:
    print('Tente novamente mais tarde')
```
