```python
# **1. Menu de operações matemáticas**

# Peça dois números e use `match-case` para escolher:

# - `1` → soma
# - `2` → subtração
# - `3` → multiplicação
# - `4` → divisão

number_1 = float(input('Digite o primeiro valor: '))
number_2 = float(input('Digite o segundo valor: '))
opcao = int(input('Escolha a opção: 1 → soma 2 → subtração 3 → multiplicação 4 → divisão'))

match(opcao):
    case _ if opcao == 1:
        print(f'A soma é: {(number_1 + number_2):.2f}')
    case _ if opcao == 2:
        print(f'A subtração é: {(number_1 - number_2):.2f}')
    case _ if opcao == 3:
        print(f'A multiplicação é: {(number_1 * number_2):.2f}')
    case _ if opcao == 4:
        print(f'A divisão é: {(number_1 / number_2):.2f}')
```

```python
# **2. Dias da semana**

# Receba um número de **1 a 7** e imprima o dia correspondente (1 = segunda, 7 = domingo).

dia_da_semana = int(input('Digite o dia da semana: '))
match(dia_da_semana):
    case 1:
        print('Segunda-feira')
    case 2:
        print('Terça-feira')
    case 3:
        print('Quarta-feira')
    case 4:
        print('Quinta-feira')
    case 5:
        print('Sexta-feira')
    case 6:
        print('Sábado')
    case 7:
        print('Domingo')
```

```python
# **3. Classificação de produto**

# Receba um código de produto e mostre sua categoria:

# - `100` → Eletrônicos
# - `200` → Vestuário
# - `300` → Alimentos

codigo_produto = int(input('Digite o código do produto: '))

match(codigo_produto):
    case 100:
        print('Categoria: Eletrônicos')
    case 200:
        print('Categoria: Vestuários')
    case 300:
        print('Categoria: Alimentos')
```

```python
# **5. Cálculo de frete**

# Receba o tipo de transporte:

# - `1` → Caminhão
# - `2` → Van
# - `3` → Moto
    
#     E exiba o valor fixo do frete.

opcao_de_frete = int(input('Escolha a opção de frete: 1 → Caminhão; 2 → Van; 3 → Moto'))
caminhao = 600
van = 300
moto = 100


match(opcao_de_frete):
    case 1:
        print('A opção escolhida foi: Caminhão')
        print(f'Frete: R${caminhao:.2f}')
    case 2:
        print('A opção escolhida foi: Van')
        print(f'Frete: R${van:.2f}')
    case 3:
        print('A opção escolhida foi: Moto')
        print(f'Frete: R${moto:.2f}')

```

```python
# **6. Conversão de moedas**

# Receba a moeda de destino (`USD`, `EUR`, `JPY`) e o valor em reais. Converta com taxas fictícias.

opcao_de_conversao = int(input('Escolha uma opção: 1 → USD; 2 → EUR; 3 → JPY'))
valor_para_converter = float(input('Digite o valor para conversão: '))

dolar = valor_para_converter / 5.42
eur = valor_para_converter / 6.35
jpy = valor_para_converter / 0.037

match(opcao_de_conversao):
    case 1:
        print(f'${dolar:.2f} ')
    case 2:
        print(f'€{eur:.2f} ')
    case 3:
        print(f'¥{jpy:.2f} ')
```

```python
# **7. Classificação de nota escolar**

# Receba uma nota de **0 a 10** e use faixas:

# - `0–4` → Reprovado
# - `5–6` → Recuperação
# - `7–10` → Aprovado

nota = float(input('Digite sua nota: '))

match(nota):
    case _ if nota > 0 and nota < 4:
        print('Reprovado')
    case _ if nota >= 4 and nota < 7:
        print('Recuperação')
    case _ if nota >= 7 and nota < 10:
        print('Aprovado')
    case _:
        print('Digite um valor válido')
```
